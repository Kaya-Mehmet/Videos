<u>***Steam***</u>

- **Problem:** Recommender systems that rely mainly on shared tags or genres can suggest games a user is unlikely to enjoy, because they ignore how the user actually plays. This project investigates whether combining user behaviour features (genre mix, log-scaled playtime, recency of play, and preference for older or newer titles) with game features (genre, tags, release age, review sentiment and price) predicts a user's enjoyment of an unseen game more accurately than tag overlap alone.

**Task:** Binary classification of whether a user recommends a game, evaluated with AUC, F1 and Precision@10, using a time-based train/validation/test split.

**Hypotheses:** (H1) Multi-feature models outperform a tag-overlap baseline. (H2) Log-scaled playtime is a more reliable engagement signal than raw hours.

**Supporting analysis:** Clustering of users (k-means/GMM with PCA) to identify player types, and feature ablation to show which factors matter most.

<u>***Anime***</u>

- Ratings are often inflated and clustered at the top of the scale, so a single numeric score says little about how much a viewer actually enjoyed a show. This project investigates whether the text of user reviews predicts a viewer's enjoyment more accurately than a model using the overall score and metadata alone. We frame this as a supervised learning task on MyAnimeList-style review data and compare text-based models from the module against a score-only baseline.

<u>***Game Popularity***</u>

- Content creation can inflate or hide a game's real popularity: some games dominate streaming while few people play them, and others have large player bases but little coverage. This project investigates whether a game's player-base popularity over time can be predicted from content-creation metrics and game attributes, and whether the gap between predicted and actual popularity identifies games that are over- or under-represented in content. We compare supervised models from the module against a simple baseline.
